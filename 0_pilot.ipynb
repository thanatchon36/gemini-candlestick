```python
# Open the images
images = [Image.open(path) for path in image_paths]

# Calculate the width and height of the merged image
widths, heights = zip(*(i.size for i in images))
total_width = sum(widths)
max_height = max(heights)

# Create a new image with the calculated size and paste the individual images
merged_image = Image.new('RGB', (total_width, max_height))
x_offset = 0
for image in images:
    merged_image.paste(image, (x_offset, 0))
    x_offset += image.size[0]

# Save the merged image to a BytesIO object
merged_image_bytes = io.BytesIO()
merged_image.save(merged_image_bytes, format='PNG')
merged_image_bytes.seek(0)

# Send the merged image to Telegram
url = f'https://api.telegram.org/bot{bot_token}/sendPhoto'
files = {'photo': merged_image_bytes}
data = {'chat_id': chat_id}
response = requests.post(url, data=data, files=files)
print(response.json())
```

```python
from img2pdf import convert
import os

def convert_png_to_pdf(png_dir, pdf_path):
    # Get a list of all PNG files in the directory
    png_files = [os.path.join(png_dir, file) for file in os.listdir(png_dir) if file.endswith('.png')]

    # Convert the PNG files to PDF
    with open(pdf_path, "wb") as pdf_file:
        pdf_bytes = convert(png_files)
        pdf_file.write(pdf_bytes)

    print(f"PDF file created: {pdf_path}")
```

```python
convert_png_to_pdf('png', 'png.pdf')
```

```python
from pdf2image import convert_from_path
from PIL import Image

def make_pdf_uncroppable(input_pdf_path, output_pdf_path):
    """Converts a PDF to a sequence of images, creating an 'uncroppable' PDF.

    Args:
        input_pdf_path (str): The path to the input PDF file.
        output_pdf_path (str): The path where the output PDF will be saved.
    """

    images = convert_from_path(input_pdf_path)

    # If your PDF has multiple pages:
    if len(images) > 1:
        images[0].save(
            output_pdf_path,
            "PDF",
            resolution=100.0,
            save_all=True,
            append_images=images[1:],
        )
    else:
        images[0].save(output_pdf_path, "PDF", resolution=100.0)

# Example usage:
input_pdf = "output.pdf"
output_pdf = "output_2.pdf"
make_pdf_uncroppable(input_pdf, output_pdf)
```

```python
from PyPDF2 import PdfMerger

def merge_pdfs(paths, output_filename):

    merger = PdfMerger()

    for path in paths:
        merger.append(path)

    merger.write(output_filename)
    merger.close()

# Example usage
pdf_paths = ["output_2.pdf", "png.pdf"]
output_file = "merged.pdf"
merge_pdfs(pdf_paths, output_file)
```
